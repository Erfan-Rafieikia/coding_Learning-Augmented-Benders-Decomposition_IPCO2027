# LABD publication analysis — version 5

This notebook reads the original `Aggregate_of_experiments.xlsx` workbook from the project `RESULTS` directory and rebuilds all publication figures, tables, LaTeX files, diagnostics, and source-data exports under `RESULTS/outputs_for_paper/publication_analysis_v5/`.


```python
from pathlib import Path
import json
import runpy
import pandas as pd
from IPython.display import display, Markdown

def locate_analysis_script() -> Path:
    cwd = Path.cwd().resolve()
    candidates = [
        cwd / 'publication_analysis_v5.py',
        cwd / 'src' / 'utils' / 'publication_analysis_v5.py',
    ]
    for candidate in candidates:
        if candidate.exists():
            return candidate.resolve()
    raise FileNotFoundError(
        'Could not locate publication_analysis_v5.py. Open this notebook from either '
        'the project root or the src/utils directory.'
    )

SCRIPT = locate_analysis_script()
SCRIPT_DIR = SCRIPT.parent
PROJECT_ROOT = SCRIPT_DIR.parents[1]
DATA = PROJECT_ROOT / 'RESULTS' / 'Experiments - Digital Research Alliance of Canada' / 'Aggregate_of_experiments.xlsx'
OUTPUT_DIR = PROJECT_ROOT / 'RESULTS' / 'outputs_for_paper' / 'publication_analysis_v5'
RUN_ANALYSIS = True

for required in [SCRIPT, DATA]:
    if not required.exists():
        raise FileNotFoundError(required)
print(f'Analysis script: {SCRIPT}')
print(f'Input workbook: {DATA}')
print(f'Output directory: {OUTPUT_DIR}')
```

## Rebuild all publication outputs


```python
if RUN_ANALYSIS:
    runpy.run_path(str(SCRIPT), run_name='__main__')
else:
    print('RUN_ANALYSIS is False; existing outputs will be inspected.')
```

## Audit diagnostics


```python
diagnostics_path = OUTPUT_DIR / '07_diagnostics' / 'analysis_diagnostics_v5.json'
diagnostics = json.loads(diagnostics_path.read_text())
display(pd.DataFrame({'Metric': diagnostics.keys(), 'Value': [str(v) for v in diagnostics.values()]}))
```

## Principal summary tables


```python
tables_dir = OUTPUT_DIR / '04_tables'
for title, filename in [
    ('Primary distributional summary', 'primary_summary_v5.csv'),
    ('SCFLP coverage', 'coverage_scflp_v5.csv'),
    ('SCMND coverage', 'coverage_scmnd_v5.csv'),
]:
    display(Markdown(f'### {title}'))
    display(pd.read_csv(tables_dir / filename))
```

## Appendix robustness summaries


```python
for title, filename in [
    ('Callback-policy robustness', 'callback_summary_v5.csv'),
    ('Feature-input ablation', 'feature_summary_v5.csv'),
]:
    display(Markdown(f'### {title}'))
    display(pd.read_csv(tables_dir / filename))
```

## Output folders

- `01_figures_for_overleaf`: flat set of all publication PDFs, matching the generated LaTeX filenames.
- `02_figures_by_problem`: organized copies by SCFLP/SCMND and analysis type.
- `03_figure_source_data`: source CSVs for every figure, plus organized copies.
- `04_tables`: CSV and LaTeX tables and output manifests.
- `05_latex`: Results section, numerical appendix, and table-only LaTeX files.
- `06_analysis_data`: cleaned, primary, callback, and feature-analysis data.
- `07_diagnostics`: audit diagnostics.
